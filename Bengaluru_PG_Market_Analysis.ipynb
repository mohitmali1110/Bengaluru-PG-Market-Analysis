## **Data Cleaning**

### We remove unnecessary columns that are not required for our business analysis

```python
df.drop(columns=['Zone','Sharing_Capacity','Amenities','Check_In_Date'],inplace=True)
df.columns
    
```

### We replace missing value with approapriate value according to column where no value was recorded

```python
df['Electricity_Charges']=df['Electricity_Charges'].fillna(0)
df['Water_Charges']=df['Water_Charges'].fillna(0)
df['Food_Type']=df['Food_Type'].fillna(df['Food_Type'].mode()[0])
df['Nearest_Metro']=df['Nearest_Metro'].fillna(df['Nearest_Metro'].mode()[0])
df['Nearest_IT_Park']=df['Nearest_IT_Park'].fillna(df['Nearest_IT_Park'].mode()[0])
df['Google_Rating'] = df['Google_Rating'].fillna(df['Google_Rating'].mean())
df['Review_Count'] = df['Review_Count'].fillna(df['Review_Count'].median())
df['Distance_to_Metro_km'] = df['Distance_to_Metro_km'].fillna(df['Distance_to_Metro_km'].median())
```

### We round Google Ratings to one decimal place for a consistent format

```python
df['Google_Rating'] = df['Google_Rating'].round(1)
```

### We verify the remaining missing values after performing initial cleaning

```python
df.isnull().sum()
```

### We standardize the column by replacing text values with numeric values

```python
df['Security_Deposit']=df['Security_Deposit'].replace({'Nil':0,'2 Months':0})
df['Monthly_Rent']=df['Monthly_Rent'].replace({'Rs 999999':999999,'999999/month':999999,'â\x82¹999999':999999,'9,99,999':'999999',-9999:9999})
```

### We convert binary amenity values into user-friendly 'Yes' and 'No' labels

```python
mapping={'0':'No',
         'NO':'No',
         'yes':'Yes',
         'Y':'Yes',
         'no':'No', 
         'False':'No',
         'True':'Yes', 
         'YES':'Yes',
         'yes':'Yes', 
         'N':'No',
         '1':'Yes'}
df[['WiFi','Laundry','Parking','Power_Backup','Housekeeping','Gym','Lift','CCTV','RO_Water','AC','TV','Study_Table','Geyser','Washing_Machine']] = df[['WiFi','Laundry','Parking','Power_Backup','Housekeeping','Gym','Lift','CCTV','RO_Water','AC','TV','Study_Table','Geyser','Washing_Machine']].replace(mapping)
```

### We convert the columns into an integer data type for calculations

```python
df['Security_Deposit']=df['Security_Deposit'].astype('int64')
df['Electricity_Charges']=df['Electricity_Charges'].astype('int64')
df['Water_Charges']=df['Water_Charges'].astype('int64')
df['Monthly_Rent']=df['Monthly_Rent'].astype('int64')
df['Review_Count'] = df['Review_Count'].astype('int64')
df.dtypes
```

### We import the Regular Expression library to perform advanced text cleaning

```python
import re

df['Area'] = df['Area'].str.lower().str.replace(
    r'\b[a-z]',
    lambda x: x.group().upper(),
    regex=True
)
df['Area'] = df['Area'].str.strip()
```

### We fill missing Area names using the most frequently occurring locality

```python
df['Area'] = df['Area'].fillna(df['Area'].mode()[0])
```

### We convert the Last Updated column into a standard datetime format

```python
df['Last_Updated'] = pd.to_datetime(
    df['Last_Updated'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
```

### Outlier Detection

```python
columns = [
    'Monthly_Rent',
    'Security_Deposit',
    'Electricity_Charges',
    'Water_Charges',
    'Maintenance_Fee'
]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

```python
plt.figure(figsize=(15,10))

for i, col in enumerate(columns, 1):
    plt.subplot(2,4,i)
    plt.boxplot(df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

#### Observation: Outliers were detected using the IQR method. However, they were not removed because most represented genuine variations in the Bengaluru PG market, such as premium rentals or higher security deposit.

## **Exploratory Data Analysis and Questions**

### Now we will read the cleaned CSV file and apply EDA on it

```python
df = pd.read_csv('../data/cleaned_bengaluru_pg_dataset.csv')
df.head()
```

### Q1. Which Bengaluru localities have the highest average monthly PG rent?

```python
avg_rent = df.groupby('Area')['Monthly_Rent'].mean().sort_values(ascending=False)

print(avg_rent.head(10))

plt.figure(figsize=(10,5))
avg_rent.head(10).plot(kind='bar')
plt.title("Top 10 Localities by Average Monthly Rent")
plt.xlabel("Area")
plt.ylabel("Average Monthly Rent (₹)")
plt.xticks(rotation=45)
plt.savefig("../images/highest_rent.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation:
- HSR Layout has the highest average monthly PG rent at approximately ₹41,829, followed by Lavelle Road at around ₹37,386.
- Premium business and IT-centric areas generally have higher PG rents due to better connectivity, employment opportunities, and access to amenities

### Q2. How does monthly rent vary across different room types?

```python
room_count = df['Room_Type'].value_counts()

print(room_count)

plt.figure(figsize=(7,7))

plt.pie(
    room_count,
    labels=room_count.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0.05]*len(room_count)
)

plt.title("Distribution of Room Types in Bengaluru PGs")
plt.axis('equal')      # Makes the pie chart circular
plt.savefig("../images/room_type.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation:
- The pie chart shows that the distribution of room types is almost evenly balanced across the dataset.
- Double rooms have the highest share with 6,200 PGs (20.6%), making them the most common room type.

### Q3. Is there a relationship between PG rent and customer ratings?

```python
correlation = df[['Monthly_Rent','Google_Rating']].corr()

print(correlation)

plt.figure(figsize=(7,5))
plt.scatter(df['Monthly_Rent'], df['Google_Rating'], alpha=0.5)
plt.title("Monthly Rent vs Google Rating")
plt.xlabel("Monthly Rent (₹)")
plt.ylabel("Google Rating")
plt.savefig("../images/google_rating_vs_rating.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation
- The scatter plot shows no clear relationship between monthly rent and Google Rating.
- PGs with both low and high rents receive similar ratings, mostly between 3.0 and 5.0.

### Q4. How does the distance to the nearest metro station influence PG rent?

```python
corr = df[['Distance_to_Metro_km','Monthly_Rent']].corr()

print(corr)

plt.figure(figsize=(7,5))
plt.scatter(df['Distance_to_Metro_km'], df['Monthly_Rent'], alpha=0.5)
plt.title("Distance to Metro vs Monthly Rent")
plt.xlabel("Distance to Metro (km)")
plt.ylabel("Monthly Rent (₹)")
plt.show()
```

#### Observation
- Monthly rent is spread across all metro distances without a clear trend.
- Premium-priced PGs are found at different distances from the metro.

### Q5. Does proximity to IT parks affect PG occupancy levels?

```python
df['IT_Distance_Group'] = pd.cut(
    df['Distance_to_IT_Park_km'],
    bins=5
)

occupancy = pd.crosstab(
    df['IT_Distance_Group'],
    df['Occupancy_Status']
)

print(occupancy)

occupancy.plot(
    kind='bar',
    stacked=True,
    figsize=(10,5)
)

plt.title("Occupancy Status by Distance to IT Park")
plt.xlabel("Distance to IT Park")
plt.ylabel("Number of PGs")
plt.xticks(rotation=45)
plt.legend(title="Occupancy")
plt.tight_layout()
plt.show()
```

#### Observation
- All distance groups have a similar distribution of occupancy status.
- The numbers of Almost Full, Available, and Full PGs remain fairly consistent across all IT park distance ranges.

### Q6. Which amenities are most commonly available in Bengaluru PGs?

```python
amenities = [
    'WiFi', 'Laundry', 'Parking', 'Power_Backup',
    'Housekeeping', 'Gym', 'Lift', 'CCTV',
    'RO_Water', 'AC', 'TV', 'Study_Table',
    'Geyser', 'Washing_Machine'
]

amenity_count = (df[amenities] == 'Yes').sum().sort_values(ascending=False)

print(amenity_count)

plt.figure(figsize=(10,6))
amenity_count.plot(kind='bar', color='skyblue')
plt.title("Most Common Amenities in Bengaluru PGs")
plt.xlabel("Amenities")
plt.ylabel("Number of PGs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../images/amenities_distribution.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

#### Observation
- Laundry (15,245 PGs) is the most commonly available amenity.
- Housekeeping and RO Water are also provided by a large number of PGs.
- Power Backup (14,936 PGs) is the least common among the listed amenities.
- Overall, the availability of amenities is quite uniform across PGs.

### Q7. How does customer satisfaction vary across different localities?

```python
plt.figure(figsize=(14,6))

sns.boxplot(
    x='Area',
    y='Google_Rating',
    data=df
)

plt.title("Distribution of Google Ratings Across Different Areas")
plt.xlabel("Area")
plt.ylabel("Google Rating")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Q8. Which localities have the highest concentration of highly rated PGs?

#### Observation

### Q9. How do additional charges (electricity, water, and maintenance) contribute to the overall cost of living?

```python
df['Additional_Charges'] = (
    df['Electricity_Charges'] +
    df['Water_Charges'] +
    df['Maintenance_Fee']
)

print(df[['Electricity_Charges',
          'Water_Charges',
          'Maintenance_Fee',
          'Additional_Charges']].describe())

average = [
    df['Electricity_Charges'].mean(),
    df['Water_Charges'].mean(),
    df['Maintenance_Fee'].mean()
]

labels = [
    'Electricity',
    'Water',
    'Maintenance'
]

plt.figure(figsize=(7,5))
plt.bar(labels, average)
plt.title("Average Additional Charges")
plt.ylabel("Average Amount (₹)")
plt.show()
```

#### Observation
- Electricity charges have the highest average cost (approximately ₹965).
- Maintenance charges average around ₹503.
- Water charges are the lowest, averaging about ₹244.
- Electricity contributes the largest portion of additional monthly expenses.

### Q13. Is there a correlation between safety score, commute score, value-for-money score, and customer ratings?

### Q14. Which localities offer the best combination of affordability, high ratings, and good accessibility?

```python
summary=df.groupby('Area')[[
'Monthly_Rent',
'Google_Rating'
]].mean()

plt.figure(figsize=(8,6))

plt.scatter(
    summary['Monthly_Rent'],
    summary['Google_Rating']
)

plt.xlabel("Average Monthly Rent")
plt.ylabel("Average Google Rating")
plt.title("Rent vs Rating Across Localities")

plt.show()
```

#### Observation
- Most localities have average rents between ₹12,000 and ₹20,000 with ratings around 3.7–3.9.
- A few premium localities have much higher rents (above ₹35,000) but only slightly higher ratings.
- Higher rent does not necessarily result in significantly better ratings.

### Q15. How do amenities influence occupancy and customer ratings?

```python
amenities=[
'WiFi',
'Laundry',
'Parking',
'Power_Backup',
'Housekeeping',
'Gym',
'Lift',
'CCTV',
'RO_Water',
'AC',
'TV',
'Study_Table',
'Geyser',
'Washing_Machine'
]

df['Amenity_Count']=(df[amenities]=='Yes').sum(axis=1)

plt.figure(figsize=(10,6))

sns.boxplot(
    x='Amenity_Count',
    y='Google_Rating',
    data=df
)

plt.title("Google Rating by Number of Amenities")
plt.xlabel("Number of Amenities")
plt.ylabel("Google Rating")

plt.show()
```

#### Observation
- Median Google Ratings remain fairly similar across different numbers of amenities.
- PGs with 13–14 amenities show slightly higher median ratings.
